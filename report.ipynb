## - Bakgrund till det ni har gjort.
Vårat projekt ska svara på vilken motorykel man ska ha baserat på några få variabler.
Man ger varibler för hur den ska se ut, hur mång hästkrafter, slagvolym, säten, antal cylindrar och budget. Med detta kan användaren få en liten lista på mc-modeler som kan matcha dessa variabler.

Vi har även en model för de mer vana mc förare som kanske skulle vilja sälja sin mc men vet inte vad den kan kosta, så får man ett predikterat pris från våran model. 

Detta projektet är tänkt för personer som är inne på spåret att skaffa en mc eller för någon som har och skulle vilja sälja sin mc. 


## - Huvudresultat.
### Klassificeringsmodellen: 
Resultatet för denna modell blev ett helt okej resultat men när man tittar på våra siffror och scores för det olika evaliderings metoder så hade det kunnat se bättre ut och vi anser att detta är på grund av våran mängd data samt våra variabler som vi ville använda för detta ändamålet. Ifall man hade använt fler variabler så ser man att det blev betydligt mycket bättre resultat men eftersom en ny mc förare kanske inte har kunskapen om alla de varabler så blir det svårt att svara på som användare. Därför blev våran model lite sämre på att klassificera med bekostnad på användarvänlighet. 

### Prediktionsmodellen: 
Resultatet för våran RandomForestRegression model blev inte så bra som vi hade tänkt men det är mest på grund av att vårat dataset inte var så stort. Pågrund av detta så predikterar den bäst på äldre mc-modeller <2020 jämfört med mc-modeller >2020. Detta är endast för att majoriteten av dessa mc-modeller är under 100k SEK. alltså allt över 2020 blir snabbt dyrare och med det sagt så presterar rfr-modellen sämre.

### Data
Datan vi använde oss av kommer från kaggle: https://www.kaggle.com/datasets/peshimaammuzammil/2023-bike-model-dataset-all-data-you-need?select=bikes_data.csv

här är alla datapunkter vi hade från datasetet:
- Company
- Country of Origin
- Model
- Number of cc
- Horsepower
- Torque
- Transmission Type
- Drivetrain
- Number of Seating
- Year
- Looks
- Body Type
- Engine Type
- Number of Cylinders
- Price (SEK)

Vi skapade en egen column bara för classificerings modellen som heter: Usage Type
För rfr så användes alla förutom Usage Type medans för rfc så användes några fåtal clumner så som:
- Price
- Looks
- Horsepower
- Number of cc
- Number of Seating

För de rader med missing values har vi droppat i processen men inte tagit bort från ursprungliga csv filen. 

Onehotencode har vi använd för att kunna ha nytta av variabler som är i datatypen string. 

Precis som vi har gjort i hela kursen så splittar vi datan i train test val. 

Hela datasetet fick vi jobba en del på eftersom till en början så var priset i indiska rupies som vi fick konvertera till SEK. Drivetrain, Transmission och BodyType fick vi också jobba på eftersom det fanns många som var samma värde men det hette olika. Till exempel för transmission:
6-speed manual, six-speed manual, six-speed-manual.
dessa är samma sak så vi fick normalisera hela datasetet för de olika nämnda columner.
